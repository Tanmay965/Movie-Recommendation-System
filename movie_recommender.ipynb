### Convert string dictionary columns to lists and extract generes and keywords

```python
import ast
```

```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

### From cast extract top 3 actors

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies['cast'].head()
```

### From cast extract director

```python
import ast

def fetch_director(text):
    L = []

    if text is None:
        return []

    if isinstance(text, str):
        try:
            text = ast.literal_eval(text)
        except:
            return []

    if not isinstance(text, list):
        return []

    for i in text:
        if isinstance(i, dict) and i.get('job') == 'Director':
            L.append(i.get('name'))
            break

    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies[['title','crew']].head()
```

### Remove Spaces From Words

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.head()
```

### Create the tags column to combine all in one 

### we cant combine a string in lists so we will convert the string overview to list overview

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['movie_id','title','tags']]
```

```python
new_df.head()
```

### Convert tags list to tags string.

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_df['tags'].iloc[0]
```

### Stemming

```python
pip install nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

### Vectorization (Convert Text → Numbers)

```python
pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
vectors = tfidf.fit_transform(new_df['tags']).toarray()
```

### check vector shape

```python
vectors.shape
```

### Calculate Movie Similarity (Cosine similarity)

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

### check similarity matrix

```python
similarity.shape
```

### Create Recommendation Function

```python
def recommend(movie):
    if movie not in new_df['title'].values:
        print("Check spelling again")
        return
         
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
recommend('Titanic')
```

```python
recommend('Avatar')
recommend('Titanic')
recommend('Batman Begins')
```

```python
recommend('Cinderella')
```

```python
recommend('Cinderella')
```

# MOVIE POPULARITY PREDICTION

```python
movies2 = pd.read_csv("tmdb_5000_movies.csv")
```

```python
movies2.columns
```

### Handle missing values

```python
new_df2 = movies2[['title','vote_count', 'vote_average', 'budget', 'runtime', 'popularity']]
```

```python
new_df2 = new_df2.fillna(0)
```

```python
new_df2.head()
```

### split data

```python
X = new_df2[['vote_count', 'vote_average', 'budget', 'runtime']]
y = new_df2['popularity']
```

### Train model

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X, y)
```

### Prediction function

```python
def predict_movie_popularity(movie_name):
    movie = new_df2[new_df2['title'] == movie_name]

    if movie.empty:
        return "Movie not found"

    features = movie[['vote_count','vote_average','budget','runtime']]
    
    prediction = lr.predict(features)

    if prediction <= 50:
        level = "Low"
    elif prediction <=150:
        level = "Medium"
    else:
        level = "High"
        
    return prediction, level
```

```python
predict_movie_popularity("Avatar")
```

```python
predict_movie_popularity("Cinderella")
```
